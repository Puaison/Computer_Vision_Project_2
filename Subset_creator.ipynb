```python
import os
import shutil
import numpy as np
import pandas as pd

dataset_path_on_drive = '/content/drive/MyDrive/Computer Vision/Project/RRDataset_test/RRDataset_final'
output_path_on_drive = '/content/drive/MyDrive/Computer Vision/Project/RRDataset_subset_final'
output_path_archive_on_drive = '/content/drive/MyDrive/Computer Vision/Project/RRDataset_subset_final.tar'

category_transformations = ["original", "redigital", "transfer"]
classes = {
  "ai": 1,
  "real": 0
}

"""
In order to make a subset balanced split "across real/fake classes and transformation categories",
we need to divide the images in 6 big families and, for each family, we gather 1500 Images:
- original/ai
- original/real
- redigital/ai
- redigital/real
- transfer/ai
- transfer/real
"""

num_images = 1500 #Number of Images taken for each big family (see above)
dataframe_template = []

for category_transformation in category_transformations:
  for class_name, label in classes.items():

    source_folder_big_family = os.path.join(dataset_path_on_drive,category_transformation,class_name)

    subset_folder_big_family = os.path.join(output_path_on_drive,category_transformation,class_name)

    os.makedirs(subset_folder_big_family, exist_ok=True)

    # It takes the files in the same order position in the folder
    image_files = [
      image_filename
      for image_filename in os.listdir(source_folder_big_family)
      if os.path.isfile(os.path.join(source_folder_big_family, image_filename))]

    # 1500 equidistant positions:
    # The first is 0, the last len(image_files)-1
    indices = np.linspace(
      0,
      len(image_files) - 1,
      num_images,
      dtype=int)

    for index in indices:
      image_filename = image_files[index]

      source_file_on_drive = os.path.join(source_folder_big_family,image_filename)

      destination_file_on_drive = os.path.join(subset_folder_big_family,image_filename)

      shutil.copy2(source_file_on_drive, destination_file_on_drive)

      dataframe_template.append({
        "path_on_drive": destination_file_on_drive,
        "category_transformation": category_transformation,
        "class": class_name,
        "label_AI": label
      })

    print(category_transformation,"/",class_name,": ",len(indices),"Selected Images")

# Create the .csv file
dataframe = pd.DataFrame(dataframe_template)

csv_path_on_drive = os.path.join(output_path_on_drive, "subset_list.csv")
dataframe.to_csv(csv_path_on_drive, index=False)

print("Total number of Images in the new generated subset of RRDataset: ",len(dataframe))
```
